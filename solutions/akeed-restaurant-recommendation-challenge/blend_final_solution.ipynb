```python
import numpy as np
import pandas as pd 
```

```python
df = pd.read_csv('akeed_colab_v93.csv')[['CID X LOC_NUM X VENDOR']]
df['A'] = pd.read_csv('preds_v4.csv').set_index('CID X LOC_NUM X VENDOR').reindex(df['CID X LOC_NUM X VENDOR'])['pred'].values
df['B'] = pd.read_csv('preds_v5.csv').set_index('CID X LOC_NUM X VENDOR').reindex(df['CID X LOC_NUM X VENDOR'])['pred'].values
df['C'] = pd.read_csv('preds_v6.csv').set_index('CID X LOC_NUM X VENDOR').reindex(df['CID X LOC_NUM X VENDOR'])['pred'].values
df['D'] = pd.read_csv('akeed_colab_v93_preds.csv')['prob_target']
```

```python
df[['A', 'B', 'C', 'D']].corr()
```

```python
df['sourabh'] = (df[['A', 'C']].mean(axis = 1) * 0.55 + df['B'] * 0.45)
df['nikhil'] = (df['D'])
np.corrcoef(df['sourabh'], df['nikhil'])
```

```python
sub = df[['CID X LOC_NUM X VENDOR']]
sub['target'] = (df['nikhil'] * 0.4 + df['sourabh'] * 0.6)
```

```python
sub['target'] = (sub['target'] > 0.125)*1
```

```python
sub['target'].value_counts(normalize = True)
```

```python
sub.to_csv('final_sub.csv', index=False)
```
