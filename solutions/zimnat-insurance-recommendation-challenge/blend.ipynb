```python
import numpy as np
import pandas as pd
```

```python
sub = pd.read_csv('m10_v2.csv')
```

```python
m10_v2 = pd.read_csv('m10_v2.csv')['Label']
m10_v4 = pd.read_csv('m10_v4.csv')['Label']
m10_v5 = pd.read_csv('m10_v5.csv')['Label']
m11_v3 = pd.read_csv('m11_v3.csv')['Label']
m14_v2 = pd.read_csv('m14_v2.csv')['Label']
m14_v4 = pd.read_csv('m14_v4.csv')['Label']
m14_v5 = pd.read_csv('m14_v5.csv')['Label']
m14_v7 = pd.read_csv('m14_v7.csv')['Label']
```

```python
enc_all_lgb_v1 = pd.read_csv('enc_all_lgb_v1.csv')['Label']
```

```python
bcl_ns = pd.read_csv('blend_unscaled_15fs_2.csv')
bcl_ns = pd.merge(sub[['ID X PCODE']], bcl_ns, on = 'ID X PCODE', how = 'left')['Label']
```

```python
holar = pd.read_csv('holar/subm/final.csv')
holar = pd.merge(sub[['ID X PCODE']], holar, on = 'ID X PCODE', how = 'left')['pred']
```

```python
nn = (m14_v2 * 0.3 + m14_v4 * 0.35 + m14_v5 * 0.35) * 0.7 + m14_v7 * 0.3
gb = ((m10_v2 * 0.75 + m10_v4 * 0.25)* 0.75 + m10_v5 * 0.25) * 0.8 + m11_v3 * 0.2
blnd = (gb * 0.7 +  enc_all_lgb_v1 * 0.3) * 0.72 + nn * 0.28
sub['Label'] = (blnd * 0.6 + bcl_ns * 0.4) * 0.6 + holar * 0.4
```

```python
sub[['ID X PCODE', 'Label']].to_csv('blend_final.csv', index=False)
```
