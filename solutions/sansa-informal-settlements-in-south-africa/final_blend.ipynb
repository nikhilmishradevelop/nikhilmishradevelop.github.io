```python
df = pd.read_csv('sub_24.csv')
df['A'] = pd.read_csv('sub_24.csv')['Label']
df['B'] = pd.read_csv('sub_38.csv')['Label']
df['C'] = pd.read_csv('sub_39.csv')['Label']
df['D'] = pd.read_csv('sub_40.csv')['Label']
df['E'] = pd.read_csv('sub_41.csv')['Label']
```

```python
sub = df[['ID']]
sub['Label'] = df['A'] * 0.5 + (df[['B', 'C', 'D', 'E']].mean(axis=1)) * 0.5
```

```python
sub["Label"].plot(kind = 'density')
```
